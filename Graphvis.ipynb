```python
import matplotlib.pyplot as plt

xs = [1, 2, 3, 4, 5]
ys = [x**2 for x in range(5)]
plt.scatter(xs, ys);
plt.show()

```

```python
import numpy as np

X = np.array([x for x in zip(xs, ys)])

mean = X.mean()
X_centered = X-mean
U, S, Vt = np.linalg.svd(X_centered);

import matplotlib.pyplot as plt
new_xs = X_centered[:, 0]
new_ys = X_centered[:, 1]

def labeled_point(x, y, label):
	plt.scatter([x], [y])
	plt.annotate(label, (x, y), textcoords="offset points", xytext=(0, 5), ha="center")

plt.scatter(new_xs, new_ys)
labeled_point(Vt[0,0], Vt[0, 1], "main");
labeled_point(Vt[1,0], Vt[1, 1], "secondary");
plt.axhline(0)
plt.axvline(0)
plt.axis("equal")
plt.grid(True, alpha=0.3)
```
