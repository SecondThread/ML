```python
import matplotlib.pyplot as plt
import numpy as np

# Given a set of points, we can compute the covariant matrix of them\
# This matrix can be viewed as a transformation
# When it is a transformation, there are two directions (in 2 dimensions) which don't rotate (eigen vectors)
# The one which is scaled the most is most important, and from there, they become less and less important.


points = np.array([
	[5, 1],
	[3, 2],
	[2, 4],
	[3, 3],
	[7, 1],
])
plt.scatter(points[:, 0], points[:, 1])
plt.axis('equal')
plt.show()
```

```python
# first, center points

centered_points = points-points.mean(axis=0)
plt.scatter(centered_points[:, 0], centered_points[:, 1])
plt.axis('equal')
plt.show()

```

```python
import math

# calculate the covariant matrix manually
covariant = centered_points.T @ centered_points / (len(centered_points)-1)

angles = np.linspace(0, 2*math.pi-.1, 100)
points = np.array([[math.cos(a), math.sin(a)] for a in angles]).T
colors = np.linspace(0, 1, len(points))


transformed_points = covariant @ points

plt.scatter(points[0], points[1], c=np.linspace(0, 1, 100))
plt.scatter(transformed_points[0], transformed_points[1], c=np.linspace(0, 1, 100))

# the columns (not rows) of np.linalg.eig are the actual vectors
eigenvalues, eigenvectors = np.linalg.eig(covariant)

for v, size in zip(eigenvectors.T, eigenvalues):
	plt.plot([0, v[0]], [0, v[1]])

plt.axis('equal')
plt.show()
```
